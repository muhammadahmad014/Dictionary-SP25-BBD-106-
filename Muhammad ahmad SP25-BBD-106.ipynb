b) write a program to count the frequency of words in a given context

```python
text = input("Enter a sentence or paragraph: ")

# Convert to lowercase and split into words
words = text.lower().split()

# Create an empty dictionary to store word frequencies
word_freq = {}

# Count the frequency of each word
for word in words:
    word = word.strip(".,!?\"'")  # Remove punctuation
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

# Print the word frequencies
print("\nWord Frequencies:")
for word, freq in word_freq.items():
    print(f"{word}: {freq}")
```

c)implement a simple inventory system using dictionaries

```python
# Initial inventory
inventory = {
    "apple": 10,
    "peach": 5,
    "watermeolon": 8
}

# Function to display inventory
def display_inventory():
    print("\nCurrent Inventory:")
    for item, quantity in inventory.items():
        print(f"{item}: {quantity}")

# Function to add items to inventory
def add_item(item, quantity):
    if item in inventory:
        inventory[item] += quantity
    else:
        inventory[item] = quantity
    print(f"Added {quantity} {item}(s).")

# Function to remove items from inventory
def remove_item(item, quantity):
    if item in inventory and inventory[item] >= quantity:
        inventory[item] -= quantity
        print(f"Removed {quantity} {item}(s).")
        if inventory[item] == 0:
            del inventory[item]
    else:
        print("Not enough stock or item doesn't exist.")

# Sample usage
display_inventory()
add_item("apple", 3)
remove_item("peach", 2)
add_item("grape", 7)
remove_item("orange", 10)
display_inventory()
```

```python
from collections import Counter

# Sample data
orders = [
    {"order_id": 1, "customer": "Ahmad", "category": "Electronics"},
    {"order_id": 2, "customer": "Bushra", "category": "Books"},
    {"order_id": 3, "customer": "Ali", "category": "Books"},
    {"order_id": 4, "customer": "Daniel", "category": "Electronics"},
    {"order_id": 5, "customer": "Eesha", "category": "Clothing"},
    {"order_id": 6, "customer": "Fatima", "category": "Books"},
]

# Count how many times each category appears
category_counts = Counter(order["category"] for order in orders)

# Find the most common category
most_common = category_counts.most_common(1)[0]

print(f"The most popular product category is '{most_common[0]}' with {most_common[1]} orders.")
```

update product quantaties

```python
# Existing inventory
inventory = [
    {"name": "mobile", "quantity": 15, "price": 999},
    {"name": "Smartphone", "quantity": 30, "price": 499.99},
    {"name": "Headphones", "quantity": 50, "price": 79.99}
]

# Function to update quantity
def update_quantity(product_name, new_quantity):
    for product in inventory:
        if product["name"].lower() == product_name.lower():
            product["quantity"] = new_quantity
            print(f"Updated quantity for {product_name} to {new_quantity}.")
            return
    print(f"Product '{product_name}' not found.")

# Example usage
update_quantity("Laptop", 20)

# Display updated inventory
print("\nUpdated Inventory:")
for product in inventory:
    print(product)
```

calculate total inventory value 

```python
# Inventory data
inventory = [
    {"name": "mobile", "quantity": 20, "price": 799.99},
    {"name": "Smartphone", "quantity": 30, "price": 499.99},
    {"name": "Headphones", "quantity": 70, "price": 79.99}
]

# Function to calculate total inventory value
def calculate_total_value():
    total_value = sum(product["quantity"] * product["price"] for product in inventory)
    return total_value

# Display result
total = calculate_total_value()
print(f"Total Inventory Value: ${total:.2f}")
```

Generate a report of low stock items 

```python
# Inventory data
inventory = [
    {"name": "Laptop", "quantity": 20, "price": 799.99},
    {"name": "Smartphone", "quantity": 3, "price": 499.99},  # Low stock
    {"name": "Headphones", "quantity": 5, "price": 79.99},    # Low stock
    {"name": "Keyboard", "quantity": 12, "price": 49.99}
]

# Function to generate low stock report
def generate_low_stock_report(threshold=10):
    low_stock_items = []
    for product in inventory:
        if product["quantity"] <= threshold:
            low_stock_items.append(product)
    return low_stock_items

# Generate and display the low stock report
low_stock = generate_low_stock_report()

if low_stock:
    print("Low Stock Items Report:")
    for item in low_stock:
        print(f"Product: {item['name']}, Quantity: {item['quantity']}, Price: ${item['price']}")
else:
    print("No low stock items found.")
```

calculate the average grade for each student

```python
# Dictionary to store student information
students = {
    "S001": {
        "name": "muhammad",
        "grades": {
            "Math": 85,
            "Science": 90,
            "English": 78,
            "History": 88
        }
    },
    "S002": {
        "name": "ahmad",
        "grades": {
            "Math": 92,
            "Science": 89,
            "English": 91,
            "History": 84
        }
    }
}

# Function to calculate average grade for a student
def calculate_average_grades(grades):
    total = sum(grades.values())
    num_subjects = len(grades)
    return total / num_subjects

# Display average grade for each student
for student_id, student in students.items():
    avg_grade = calculate_average_grades(student["grades"])
    print(f"Student ID: {student_id}")
    print(f"Name: {student['name']}")
    print(f"Average Grade: {avg_grade:.2f}")
    print()
```

Find the highest and lowest grades in a specific subject :

```python
# Dictionary to store student information
students = {
    "S001": {
        "name": "Faiz rasool",
        "grades": {
            "Math": 85,
            "Science": 90,
            "English": 78,
            "History": 88
        }
    },
    "S002": {
        "name": "muhammad ahmad",
        "grades": {
            "Math": 92,
            "Science": 89,
            "English": 91,
            "History": 84
        }
    },
    "S003": {
        "name": "Ammar",
        "grades": {
            "Math": 78,
            "Science": 95,
            "English": 85,
            "History": 87
        }
    }
}

# Function to find highest and lowest grades in a specific subject
def find_highest_and_lowest(subject):
    highest_grade = None
    lowest_grade = None
    highest_student = ""
    lowest_student = ""
    
    for student_id, student in students.items():
        if subject in student["grades"]:
            grade = student["grades"][subject]
            
            # Find the highest grade
            if highest_grade is None or grade > highest_grade:
                highest_grade = grade
                highest_student = student["name"]
            
            # Find the lowest grade
            if lowest_grade is None or grade < lowest_grade:
                lowest_grade = grade
                lowest_student = student["name"]
    
    return highest_student, highest_grade, lowest_student, lowest_grade

# Example usage: Find highest and lowest grades in Math
subject = "Math"
highest_student, highest_grade, lowest_student, lowest_grade = find_highest_and_lowest(subject)

print(f"Highest grade in {subject}: {highest_student} with {highest_grade}")
print(f"Lowest grade in {subject}: {lowest_student} with {lowest_grade}")
```

sort students by their average grades:

```python
# Dictionary to store student information
students = {
    "S001": {
        "name": "muhammad ali",
        "grades": {
            "Math": 85,
            "Science": 90,
            "English": 78,
            "History": 88
        }
    },
    "S002": {
        "name": "muhammad ahmad",
        "grades": {
            "Math": 92,
            "Science": 89,
            "English": 91,
            "History": 84
        }
    },
    "S003": {
        "name": "atif ali",
        "grades": {
            "Math": 78,
            "Science": 95,
            "English": 85,
            "History": 87
        }
    }
}

# Function to calculate average grade for a student
def calculate_average(grades):
    total = sum(grades.values())
    num_subjects = len(grades)
    return total / num_subjects

# Create a list of students with their average grades
students_with_avg = [
    {"id": student_id, "name": student["name"], "average": calculate_average(student["grades"])}
    for student_id, student in students.items()
]

# Sort the list by average grade in descending order (highest first)
students_with_avg.sort(key=lambda x: x["average"], reverse=True)

# Display sorted students by average grade
print("Students sorted by their average grades:")
for student in students_with_avg:
    print(f"{student['name']}: Average Grade = {student['average']:.2f}")

```
